Пусть $X_1, X_2, \ldots, X_n$ — выборка из экспоненциального распределения с параметром $\lambda$. Найти оценку максимального правдоподобия параметра $\lambda$, сравнить ее с байесовской оценкой (MAP и математическое ожидание апостреорного распределения), подобрав сопряженное распределение. Сравнить полученные байесовские оценки с оценкой MLE. Найти предсказательное распределение

**MLE оценка**
$$L = \prod\limits_{i=1}^n\lambda e^{-\lambda x} = \lambda^n \ e^{-\lambda\sum x_i}$$
$$ln(L) = ln \ \lambda^n + ln \ e^{-\lambda\sum x_i} =   n \ ln \ \lambda - \lambda \sum x_i$$

$$\frac{\partial\ ln(L)}{∂\ λ} = \frac{n}{\lambda} -\sum x_i$$
$$\frac{n}{\lambda} = \sum x_i$$
$$\lambda_{MLE}  = \frac{n}{\sum x_i} = \frac{1}{\bar x}$$

**MAP оценка**

$$P(\lambda\ | \ D) \propto P(\lambda) \ P(D \ | \ λ)$$

Априорным сопряженным распределением у нас выступает Gamma распределение:
$$ \text{Gamma}(\alpha, \ \beta) \propto \lambda^{\alpha - 1} e^{-\lambda \ \beta} $$

$$P(\lambda \ | \ D) \propto \lambda^{\alpha - 1} e^{-\beta \ λ} \  \lambda^n \ e^{-\lambda\sum x_i} = λ^{n + α - 1} e^{-\lambda \sum x_i + \beta} \propto \text{Gamma}\left( \alpha + n, \ \beta + \sum x_i \right)$$

$$\lambda_{MAP} = \frac{\alpha + n - 1}{\beta + \sum x_i}$$
$$\mathbb{M}\left[P(\lambda\ | \ D)\right] = \frac{\alpha + n}{\beta + \sum x_i}$$





**Предсказательное распределение**

$$\mathbb P(D) = \int\limits_{λ} \mathbb P(D \ | \ \lambda) \ \mathbb P(\lambda) = \int\limits_{0}^{\infty}\lambda^n \ e^{-\lambda\sum x_i} \ \frac{\beta^{\alpha}}{\Gamma(\alpha)} \lambda^{\alpha - 1} \ e^{-\beta\lambda} d\lambda = \frac{\beta^{\alpha}}{\Gamma(\alpha)} \int\limits_{0}^{\infty}\lambda^{n + \alpha - 1} \ e^{-\lambda\left(\sum x_i + \beta\right)} =
\frac{\beta^{\alpha}}{\Gamma(\alpha)} \frac{\Gamma(\alpha + n)}{\left(\beta + \sum x_i\right)^{\alpha + n}}$$

$$\mathbb P(\lambda \ | \ D) = \frac{\lambda^n \ e^{-\lambda \sum x_i} \ \frac{\beta^{\alpha}}{\Gamma(\alpha)} \ \lambda^{\alpha - 1} \ e^{-\beta\lambda}}{\frac{\beta^{\alpha}}{\Gamma(\alpha)} \frac{\Gamma(\alpha + n)}{\left(\beta + \sum x_i\right)^{\alpha + n}}} = \frac{\left( \beta + \sum x_i \right)^{\alpha + n} \ e^{-\lambda\left(\sum x_i + \beta\right)} \ \lambda^{n + \alpha - 1}}{\Gamma(\alpha + n)}$$


$$\mathbb P(x_{new} \ | \ D) = \int\limits_0^{\infty} \lambda \ e^{-\lambda x_{new}} \frac{\left( \beta + \sum x_i \right)^{\alpha + n} \ e^{-\lambda\left(\sum x_i + \beta\right)} \ \lambda^{n + \alpha - 1}}{\Gamma(\alpha + n)} = \frac{\left( \beta + \sum x_i \right)^{\alpha + n}}{\Gamma(\alpha + n)}\int\limits_0^{\infty} \lambda^{n + \alpha} \ e^{-\lambda \left(\sum x_i + \beta + x_{new}\right)} = \frac{\left( \beta + \sum x_i \right)^{\alpha + n}}{\Gamma(\alpha + n)} \frac{\Gamma(\alpha + n + 1)}{\left(\beta + \sum x_i + x_{new}\right)^{\alpha + n + 1}} = \frac{\left( \beta + \sum x_i \right)^{\alpha + n}}{\Gamma(\alpha + n)} \frac{(\alpha + n) \ \Gamma(\alpha + n)}{\left(\beta + \sum x_i + x_{new}\right) \ \left(\beta + \sum x_i + x_{new}\right)^{\alpha + n}} = \frac{\left( \beta + \sum x_i \right)^{\alpha + n} \ (\alpha + n)}{\left(\beta + \sum x_i + x_{new}\right) \ \left(\beta + \sum x_i + x_{new}\right)^{\alpha + n}} = \frac{(\alpha + n)}{\left(\beta + \sum x_i + x_{new}\right)} \left(\frac{\beta + \sum x_i}{\beta + \sum x_i + x_{new}}\right)^{\alpha + n}$$

# Задача 2

**Мультиномиальное распределение**

Пусть проводится серия из $n$ испытаний и в результате каждого испытания происходит ровно одно событие из набора $A_1, A_2, \dots, A_m$, причем вероятности этих событий равны соответственно $\mathsf{p}_1, \mathsf{p}_2, \dots, \mathsf{p}_m$, причем
$$\sum_{i=1}^{m}\mathsf{p}_i = 1.$$

Тогда совместное распределение величин $X_1, X_2, \dots, X_m$, где $X_k$ — число наступлений события $A_k$ в серии из $n$ испытаний, задается вероятностями

$$
\mathsf{P}\left(X_1 = n_1, \dots, X_m = n_m, \right) = \frac{n!}{n_1!\dots n_m!}\mathsf{p}_1^{n_1}\dots \mathsf{p}_m^{n_m},
$$

где $n_1, n_2, \dots, n_m$ — произвольный набор целых неотрицательных чисел, таких что

$$\sum_{i=1}^m n_i = n.$$

Произведите байесовский вывод для мультиномиального распределения: найдите апостериорное распределение, используя в качестве сопоряженного распределения к правдоподобию [распределение Дирихле](https://ru.wikipedia.org/wiki/%D0%A0%D0%B0%D1%81%D0%BF%D1%80%D0%B5%D0%B4%D0%B5%D0%BB%D0%B5%D0%BD%D0%B8%D0%B5_%D0%94%D0%B8%D1%80%D0%B8%D1%85%D0%BB%D0%B5), найдите предсказательное распределение. Объясните результат.

$$\mathbb P(D \ | \ p_1, \ldots, p_m) = \frac{n!}{n_1!\dots n_m!} \prod\limits_{i=1}^m \mathsf{p}_i^{n_i}$$

$$\mathbb P(p_1, \ldots, p_m \ | \ D) \propto \mathbb P(D \ | \ p_1, \ldots, p_m) \ \mathbb P(p_1, \ldots, p_m) = \prod\limits_{i=1}^m \mathsf{p}_i^{n_i} \prod\limits_{i=1}^m \mathsf{p}_i^{\alpha_i - 1} = \prod\limits_{i=1}^m \mathsf{p}_i^{n_i + \alpha_i- 1}$$

$$\mathbb P(D) = \int\limits_p \mathbb P(D \ | \ p) \ \mathbb P(p) = \int\limits_p \frac{n!}{n_1!\dots n_m!} \prod\limits_{i=1}^m \mathsf{p}_i^{n_i}\frac{1}{Β(\alpha)}\prod\limits_{i=1}^m \mathsf{p}_i^{\alpha_i - 1} = \frac{n!}{n_1!\dots n_m! \ Β(\alpha)} \int\limits_p \prod\limits_{i=1}^m \mathsf{p}_i^{n_i + \alpha_i - 1} = \frac{n! \ Β(\alpha + n)}{n_1!\dots n_m! \ Β(\alpha)}$$

$$\mathbb P(x_{new} = k \ | \ D) = \int\mathbb P(x_{new} = k \ | \ p) \ \mathbb P(p \ | \ D) = \int p_k \ \mathbb P(p \ | \ D) = \int p_k \frac{\frac{\prod\limits_{i=1}^m \mathsf{p}_i^{n_i + \alpha_i- 1}}{\frac{n!}{n_1!\dots n_m!\ Β(\alpha)}}}{\frac{n!}{n_1!\dots n_m!\ Β(\alpha)} Β(\alpha + n)} = \int p_k \frac{\prod\limits_{i=1}^m \mathsf{p}_i^{n_i + \alpha_i- 1}}{Β(\alpha + n)} = \frac{1}{Β(\alpha + n)} \int p_k \prod\limits_{i=1}^m \mathsf{p}_i^{n_i + \alpha_i- 1} = \frac{1}{Β(\alpha + n)} \int \prod\limits_{i=1}^m \mathsf{p}_i^{n_i + \alpha_i} =  \frac{Β(\alpha + n + 1_k)}{Β(\alpha + n)} = \frac{\prod\limits_{i=1}^m \Gamma(\alpha_i + n_i + 1_k)}{\Gamma\left(\sum\limits_{i=1}^m\alpha_i + n_i + 1_k\right)} \frac{\Gamma\left(\sum\limits_{i=1}^m\alpha_i + n_i \right)}{\prod\limits_{i=1}^m \Gamma(\alpha_i + n_i )} = \frac{(\alpha_k + n_k)}{\sum\limits_{i=1}^m\alpha_i + n_i}$$
